The most important hardware component in a computer is the central processing unit, or *CPU*. This is the chip that performs the actual calculations. At the lowest level, a CPU can request data from memory and perform simple hardware operations on basic datatypes such as floats and integers. These basic operations include comparison, addition, subtraction, multiplication, division, and taking square roots. More complex operations, such as calculating a power, a logarithm or a cosine, are implemented in software, in terms of these simple hardware operations.

It requires even more steps to translate your Python code into something that can be executed in terms of simple hardware operations. The initial steps are executed by the Python *interpreter*. Simply put, the interpreter is a program that determines what your Python code means and executes it. Consider the code `x ** 2.4`. To evaluate this, the Python interpreter first needs to look up `x`. If `x` is a built-in numeric type such as `int` or `float`, it will call a math software library that can calculate the power. This math software library has algorithms to calculate it in terms of hardware operations. If `x` is an object of a custom class, the Python interpreter may call its `__pow__()` method and take some more steps before the actual operation is executed on the CPU.

Because Python code is evaluated by the Python interpreter, it is called an *interpreted language*. Other interpreted languages are e.g. [MATLAB](https://nl.mathworks.com/products/matlab.html) and [R](https://www.r-project.org/). Interpreted languages have the advantage that written code can be executed immediately, that changes to e.g. functions and classes can be tested immediately, and that programs can run on different operating systems. They also have the advantage that datatypes are determined dynamically at runtime; in the example above, `x` can be an integer, a float, or even a NumPy array. Dynamic datatypes can make functions more generic.

*Compiled languages*, on the other hand, first translate written code into machine code. This translation is done by a *compiler*. The compiled machine code can be executed on the CPU and can directly call math libraries, operating system libraries, etc. Once compiled, machine code no longer needs to be interpreted, and programs execute much faster. But compilation has to be done for a specific *operating system* and a specific *CPU architecture*. Many compiled languages also require strict typing; the compiler has to know beforehand if the variable `x` is a 16-bit unsigned integer, or a 32-bit floating point. Examples of compiled languages are [C](https://en.wikipedia.org/wiki/C_(programming_language)), [C++](https://en.wikipedia.org/wiki/C%2B%2B), and [Fortran](https://en.wikipedia.org/wiki/Fortran).

Often, interpreted languages are also *higher level languages* than compiled languages, because things like *memory management* are abstracted away. They allow you to write more generic programs with fewer lines of code. The interpreter program itself is usually written in a compiled language: the reference Python interpreter, [CPython, is written in C](https://github.com/python/cpython/tree/main/Python).

Below are examples of a `Circle` class implemented in Python, MATLAB, and C++. Although the syntax slightly differs between the three languages, the overall structure is the same. Can you read and understand all three implementations? Once you understand the fundamental concepts of object-oriented programming, learning a new programming language mainly involves getting familiar with its syntax.

## Python

```python
import math


class Circle:
    def __init__(self, radius):
        self._radius = radius

    def area(self):
        return math.pi * self._radius**2

    def circumference(self):
        return 2 * math.pi * self._radius

    def display(self):
        print(f"Radius: {self._radius}")
        print(f"Area: {self.area()}")
        print(f"Circumference: {self.circumference()}")
```

## MATLAB

```matlab
classdef Circle
    properties (Access = private)
        radius
    end

    methods
        function obj = Circle(radius)
            obj.radius = radius;
        end

        function a = area(obj)
            a = pi * obj.radius^2;
        end

        function c = circumference(obj)
            c = 2 * pi * obj.radius;
        end

        function display(obj)
            disp("Radius: " + obj.radius);
            disp("Area: " + obj.area());
            disp("Circumference: " + obj.circumference());
        end
    end
end
```

## C++

```cpp
#include <iostream>


constexpr double pi = 3.141592653589793;


class Circle {
private:
    double radius;

public:
    Circle(double r) {
        radius = r;
    }

    double area() const {
        return pi * radius * radius;
    }

    double circumference() const {
        return 2 * pi * radius;
    }

    void display() const {
        std::cout << "Radius: " << radius << '\n';
        std::cout << "Area: " << area() << '\n';
        std::cout << "Circumference: " << circumference() << '\n';
    }
};
```
