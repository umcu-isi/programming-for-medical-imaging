We've come across quite some different datatypes in past three weeks, such a `string`, `datetime`, `np.ndarray`, `pydicom.dataset.Dataset`, etc. You probably have noticed that most of these datatypes are not just containers for data, but that they also have functions to work *with* this data.

Such datatypes, with member variables and member functions, are called *classes*. An instance of such a datatype is called an *object*.

Member variables are also called *class attributes*, or simply *attributes*. Member functions are known as *class methods*, or simply *methods*, in Python. A Numpy array object, has e.g. an attribute `shape` and methods to calculate the sum and the mean. It also knows how to display itself when you call `print()`:

```python
import numpy as np


identity = np.eye(3) + 9
print('shape:', identity.shape)  # Get the value of the 'shape' attribute.
print('sum: ', identity.sum())  # Calculate the sum of all elements.
print('mean: ', identity.mean())  # Calculate the mean of all elements.
print(identity)  # Knows how to print an array.
```

Similar, a pydicom `Dataset` object has methods to add DICOM attributes, and these DICOM attributes are objects too, with properties of their own, such as a tag number:

```python
from pydicom import Dataset


ds = Dataset()
ds.PatientName = "Cleese^John"
ds.Modality = "CT"

print(ds)  # Knows how to print the DICOM attributes
print('The attribute PatientName is an object of the type:', type(ds.PatientName))  # Is an object of the class pydicom.valuerep.PersonName
```

Programming with classes and object is called *object-oriented programming*, or *OOP*. The main advantage of object-oriented programming is that it gives you a way to naturally organize complex code around objects that have both *state* (what an object is) and *behavior* (what an object does).

Let's take the list `numbers = [1, 2, 3]`. The list's state is that it contains 3 elements; the numbers 1, 2, and 3. The lists behavior is that it can do things to that state, and with that state, like: `numbers.append(4)`, `numbers.reverse()`, or `print(numbers)`. In this example `list` is the class and `numbers` is an object of that class.

Organizing code through classes also forces keeping related code together, and hides internal details of objects, keeping the API clean.


## Defining Classes


New class datatypes can be created with the keyword `class`. By convention, class names are in *'CamelCase'* (meaning *no spaces and capitalizing the first letter of every word*).

To initialize a new object to a valid state, we need at least an `__init__()` method. This method is called automatically when a new object is created.

Python uses these double underscores (also known as *dunders* or *magic methods*) to define class methods with a *special meaning*. You are not supposed to call these methods yourself, but Python (or other classes) will call such methods automatically when necessary.

Methods can be added to a class in the same way as definition regular functions, using `def`. Class methods, however, are always called with an extra argument, `self`, that refers to the current object. The argument `self` always comes first, and is automatically added when the method is called.

A simple class for bookkeeping scans for specific subjects could look like this:

```python
class Subject:
    def __init__(self, patient_id):
        self.id = patient_id  # Assign 'patient_id' to a new class attribute 'id'
        self.scans = []  # Initialize a new class attribute 'scans' as an empty list.

    def add(self, scan):  # Appends 'scan' to the attribute 'scans'.
        self.scans.append(scan)

    def get_directories(self):  # Returns the directory for each scan.
        return [f"{self.id}/{s}/" for s in self.scans]
```

Let's try out our new class:

```python
# Create 2 subjects and add scans
subjects = [Subject("007"), Subject("005")]
subjects[0].add("CT_123")
subjects[0].add("MR_456")
subjects[1].add("CT_789")

# Print the subject IDs and their scans:
for s in subjects:
    directories = s.get_directories()
    print(f'{s.id}: {directories}')

# Print the datatype of the first subject.
print(type(subjects[0]))
```
