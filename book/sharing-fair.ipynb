It is very likely that you will some day use data and software from publicly funded research, or perhaps contribute to software and data created in publicly funded research. In these cases, it is important that data and software is *Findable*, *Accessible*, *Interoperable* and *Reusable*; FAIR. The FAIR principles are a concept which originated in data management, but they are applicable to software as well.

FAIR software is about more than making code publicly available. It is about making software discoverable, (legally) reusable, attributable, and maintainable, so research can be reproduced and built upon.

> This website gives five recommendations for FAIR software: <https://fair-software.nl/>

When writing new code, keep *Interoperable* and *Reusable* in mind. Try to create readable, generic functions and classes as much as possible. Don't *hard-code* specifics for your project, but put such specifics in configuration files and treat them as variables in your code. When you decide to publish your work *open source*, meaning that your source code is publicly available, it is a good idea follow the FAIR guidelines. Besides making sure that other people have all the information they need to use your code, you should at least add a *software license*. Contrary to common belief, the main reason for adding a license is not to legally restrict use, but rather to *legally allow use* of your code. Without a license, copyright laws (such as the Dutch [Copyright Act (Auteurswet)](https://business.gov.nl/regulations/copyright/) and the [EU copyright law](https://digital-strategy.ec.europa.eu/en/policies/copyright-legislation)) prohibit copying works of others. The FAIR-software website mentioned above explains how to choose an appropriate license.

The importance of open source software goes beyond the fact that software is free of charge. It has significant benefits for research, innovation, education and society. But open source software that is not FAIR has much less of an impact than software that is.

Even when you can't publish all your work open source, it is still a good idea to follow as much of the FAIR guidelines as possible! It is just as important being FAIR to your colleagues and yourself as being FAIR to the rest of the world. 🙂
