## MicroDicom practicum
MicroDicom is a free tool to view and study medical images. These steps cover installation, basic tools, and measurements.

Getting Started

1. Download the software from the MicroDicom website (see above), install it and execute it on your computer.
2. Download the set of sample DICOM images from …
3. 2D image manipulation. Click File and select Open (or drag a file) to load the DICOM file “Knee.dcm”
    - Look for the most regular functions to manipulate images:
        * Click the Zoom icon or hold the right mouse button to make the image bigger or smaller
        * Select the Pan icon (hand symbol) to move the image within the viewport
        * Adjust the brightness and contrast by clicking the Window/Level tool and dragging your mouse over the picture
        * Create some annotations (lines, curves, circles, etc) and delete them again
        * Find out how to measure the angle of 2 lines
        * Apply different color scales, display the knee for instance in the heated object scale (HOT)
    - The image contains a metal wedge (aluminium). What could be a reason to add this device in an X-ray?
     - Each step of the wedge is 1cm high, 4cm width. Find the measurement tool and measure the length of the wedge, e.g. of 10 steps (10cm) height. With this knowledge, you should be able to calculate the pixel size of this X-ray. Show the calculation to a teaching assistant (TA), or ask for help.
     - Find the stored pixel size in the DICOM header. Compare this with the results of your distance measurement(s). What could be causes for the differences?
4. Next, drag a full stack of CT slices to MicroDicom. Choose any set you want.
    - Use your mouse wheel to scroll up and down through the image slices. Check the effect and differences if you use the arrow buttons and the PgUp PgDn buttons to scroll through the images.
    - Use the Angle tool to draw lines and find the angle between them.
    - Place the ROI (Region Of Interest) tool on a specific spot to see average pixel values.
    - Choose View->Dicom Tags Pane. This way, the DICOM header information of the current image will be shown on the right.
    - Scroll through the 3D image and try to find tags that change if you scroll.
    - You can add some tags to Favorites with your mouse (right mouse button click).
    - Open the CT pelvis and add the tags "(0020,1041)", "(0020,0032)", and "(0018,0050)" to Favorites. Make the Favorites Tags active (right below) and start scrolling again. Try to explain the changes in the values and ask for help if you don't understand it.



## ITK-SNAP practicum
ITK-SNAP is also a free tool to view and study medical images. This application van also images in Nifti format, which we will use throughout the PMI course as well.

1. Download the software from the ITK-SNAP website (see above), install it and execute it on your computer.
2. Download the set of sample Nifti images from …
3. Click File and select Open (or drag a file) to load the Nifti file “reg_T1.nii.gz” from any of the subfolders in MRBrains18.
    - Scroll in all 3 visible windows and make sure you understand what is being displayed in every viewport as you scroll. As with MicroDicom, look for the most regular functions to manipulate images:
        * Click the Zoom icon or hold the right mouse button to make the image bigger or smaller
        * Panning can be done with the middle mouse button
        * Try to find the tool that handles the contrast (window/level). Adjust the brightness and contrast and play with the available options
        * Check the Metadata of the image. Do you see the differences with the header information of DICOM images?
        * Try to find the Annotation tool, draw some lines and look at the information shown in the image.
        * Change the color of the lines.
        * Delete an annotation.
    - Load a new image "T1.nii.gz" as Main image.
        * Load "T1_mask.nii.gz" as Segmentation (drag this file into ITK-SNAP, a window will pop-up).
        * Scroll through all slices and try to think what the purpose is of the mask.
        * Unload the mask as segmentation, and create your own segmentation using the Paint brush mode. Try to find the volume of your own drawing.
