This week, we will start with explaining different file formats to store (medical) images. There are many different formats for 2D images, each with their own specific pros and cons. Medical images are often multi-dimensional (3D, 3D-time, etc), which requires specific features to store all the information. In this course, we will use the DICOM (dcm) and Nifti (nii) formats.
