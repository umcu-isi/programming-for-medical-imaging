The most important part of a computer is the central processing unit, or *CPU*. This is the chip that performs the actual calculations. A CPU can request data from memory and perform basic operations on basic datatypes such as floats and integers. A CPU can compare values, add, subtract, multiply, divide, and take square roots. It can also do Boolean operations such as *and*, *or*, *not*, etc.

Python does not directly execute your code on the CPU. It requires a few steps to translate your code into something that can be executed in terms of the basic operations that a CPU knows. These steps are executed by the Python *interpreter*. Simply put, the interpreter is a program that determines what your Python code means and executes it.

Consider the code `print(x ** 3)`. Python first evaluates `x ** 3`. To do this, it needs to look up `x`. If `x` is a numeric type, it will call a math software library that can calculate the power. This math software library has algorithms to calculate e.g. a power or a logarithm in terms of basic CPU operations. If `x` is an object of a custom class, the Python interpreter may call its `__pow__()` method and take some more steps before this math software library is involved. Because we asked to print the result, the interpreter calls the `__str__()` method to get a string that is written to the *output stream*. The *operating system* software (i.e. Windows, macOS, Linux), manages this stream and makes sure that it is displayed in a terminal, or written to a file or another destination.

Because Python code is evaluated by the Python interpreter, it is called an *interpreted language*. Other interpreted languages are e.g. [MATLAB](https://nl.mathworks.com/products/matlab.html), [R](https://www.r-project.org/), and [JavaScript](https://en.wikipedia.org/wiki/JavaScript). Interpreted languages have the advantage that written code can be executed immediately, that changes to e.g. functions and classes can be tested immediately, and that programs can run on different operating systems. They also have the advantage that datatypes are dynamically interpreted; in the example above, `x` can be an integer, a float, or even a NumPy array. Dynamic datatypes can make functions more generic.

*Compiled languages*, on the other hand, first translate written code into machine code. This translation is done by a *compiler*. The compiled machine code can be executed on the CPU and can directly call math libraries, operating system libraries, etc. Because this machine code no longer needs to be interpreted, programs execute much faster. But compilation has to be done for a specific *operating system* and a specific *CPU architecture*. And datatypes have to be strict; the compiler has to know beforehand if the variable `x` is a 16-bit unsigned integer, or a 32-bit floating point. Examples of compiled languages are [C](https://en.wikipedia.org/wiki/C_(programming_language)), [C++](https://en.wikipedia.org/wiki/C%2B%2B), and [Fortran](https://en.wikipedia.org/wiki/Fortran).

Most often, interpreted languages are also *higher level languages* than compiled languages, because things like *memory management* are abstracted away. They allow you to write more generic programs with fewer lines of code. The interpreter program itself is usually written in a compiled language: the reference Python interpreter, [CPython, is written in C](https://github.com/python/cpython/tree/main/Python).

Below is an example of a `Circle` class implemented in Python, MATLAB, R, and C++. Note that, although the *syntax* is a little different, that the overall structure is the same. Can you read and understand all four implementations? Once you know the basic concepts of (object-oriented) programming, learning a new language means getting familiar with a new syntax and a new package *ecosystem*.

::: {.panel-tabset}
## Python

```python
import math


class Circle:
    def __init__(self, radius):
        self._radius = radius

    def area(self):
        return math.pi * self._radius**2

    def circumference(self):
        return 2 * math.pi * self._radius

    def display(self):
        print(f"Radius: {self._radius}")
        print(f"Area: {self.area()}")
        print(f"Circumference: {self.circumference()}")
```

## MATLAB

```matlab
classdef Circle
    properties (Access = private)
        radius
    end

    methods
        function obj = Circle(radius)
            obj.radius = radius;
        end

        function a = area(obj)
            a = pi * obj.radius^2;
        end

        function c = circumference(obj)
            c = 2 * pi * obj.radius;
        end

        function display(obj)
            disp("Radius: " + obj.radius);
            disp("Area: " + obj.area());
            disp("Circumference: " + obj.circumference());
        end
    end
end
```

## R

```r
Circle <- R6Class("Circle",

  private = list(
    radius = NULL
  ),

  public = list(

    initialize = function(radius) {
      private$radius <- radius
    },

    area = function() {
      pi * private$radius^2
    },

    circumference = function() {
      2 * pi * private$radius
    },

    display = function() {
      cat("Radius:", private$radius, "\n")
      cat("Area:", self$area(), "\n")
      cat("Circumference:", self$circumference(), "\n")
    }
  )
)
```

## C++

```cpp
#include <iostream>


constexpr double pi = 3.141592653589793;


class Circle {
private:
    double radius;

public:
    Circle(double r) {
        radius = r;
    }

    double area() const {
        return pi * radius * radius;
    }

    double circumference() const {
        return 2 * pi * radius;
    }

    void display() const {
        std::cout << "Radius: " << radius << '\n';
        std::cout << "Area: " << area() << '\n';
        std::cout << "Circumference: " << circumference() << '\n';
    }
};
```
