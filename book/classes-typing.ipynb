In the past few weeks we've encountered many datatypes, and in the past chapters we've even created a few new datatypes by creating our own classes. We've written functions and class methods that are supposed to work with certain datatypes. For example, take the function:

```python
def add(x, y):
    return x + y
```

This function works with integers, floats, and even NumPy arrays. But it is not supposed to work with other datatypes, such as `str` or `list`.

```python
import numpy as np


print(add(2, 3.5))  # add() works with integers and floats
print(add(np.array([1, 2, 3]), np.array([4, 5, 6])))  # it also works with NumPy arrays
print(add(3.5, []))  # it does not work with lists
```

Now take this implementation. Functionally, it is the same as the implementation above, but this definition has so-called *type hints*:

```python
def add(x: float, y: float) -> float:
    return x + y
```

These type hints suggest that the input arguments to `add()` should be two floats, and that the function should also return a `float`.

Being type *hints*, this is not enforced in Python, but they are interpreted by most IDEs, like PyCharm. Running the code with another type will not raise an error.

Type hints are a good idea because they make Python code easier to understand, safer to change, and easier to maintain. They let you document what a function expects and what it returns without changing the behavior of functions. When you start using type hints in your IDE, you'll find that the IDE not only warns about type errors, but that it also gets better in showing documentation, auto-completion, and auto-generating code.

When multiple types are allowed as input, or if multiple types could be returned by a function, the pipe-symbol (`|`) can be used to separate them. The `|` should be read as 'or'. Square brackets (`[]`) can be used to specify types inside a container datatype. For example:

* `float | int` - a float or an integer
* `float | int | np.array` - a float, an integer, or a NumPy array
* `str | None` - a string or None
* `list[float | int]` - a list of floats or integers
* `list[float] | list[int]` - a list floats or a list of integers
* `tuple[float, float]` - a tuple of two (and only two) floats
* `dict[str, int]` - a dictionary mapping strings to integers

These type definitions can be assigned to a variable, making is easier to specify our type hints. Let's inform the users and IDE that our `add()` function can also support integers and NumPy arrays:

```python
Numeric = float | int | np.ndarray

def add(x: Numeric, y: Numeric) -> Numeric:
    return x + y
```

To enforce type checking during runtime, we can use the `isinstance()` function that we saw in the previous chapters. `isinstance()` returns True if the object belongs to one of the classes in the union.

```python
Numeric = float | int | np.ndarray

def add(x: Numeric, y: Numeric) -> Numeric:
    if not isinstance(x, Numeric) or not isinstance(y, Numeric):
        raise TypeError('x and y must be numeric')
    return x + y
```

Trying to add something to a string will now raise a `TypeError`:

```python
add("Brian", 42)
```

Note that we can use *any* type or class name in our type hints, including types that we defined ourselves, such as the `Module` class from the previous chapter. Let's add type hints and checks to that class:

```python
Numeric = float | int | np.ndarray


class Module:
    def __init__(self):
        print("Initializing an instance of the class", type(self).__name__)

    def forward(self, x: Numeric) -> Numeric:
        """
        Executes the forward pass of the module.
        Should be overridden by all derived classes.

        :param x: The input to the module.
        :return: The result of the computation.
        """
        raise NotImplementedError

    def __call__(self, x: Numeric) -> Numeric:
        if not isinstance(x, Numeric):
            raise TypeError('x must be numeric')

        return self.forward(x)


class Sequential(Module):
    def __init__(self, modules: list[Module]):
        if any(not isinstance(m, Module) for m in modules):
            raise TypeError('All modules must derive from Module')

        super().__init__()
        self._modules = modules

    def forward(self, x: Numeric) -> Numeric:
        # Iterate over all modules and execute them, passing the output
        # of a previous module as input to the next.
        for module in self._modules:
            x = module(x)
        return x
```

```python
model = Sequential(["double", "addone"])
```
