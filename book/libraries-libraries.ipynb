As our Python scripts and programs become larger, putting all the code in a single file quickly becomes difficult to manage. A better approach is to organize code into *modules* and *libraries*, to reuse existing code, and to separate different parts of an application into logical components.

This week's course material focuses on creating and working with *libraries*.

## Python Libraries

A *library* is a collection of code that other programs can use. There are many publicly available libraries allowing programmers to solve common problems without writing everything themselves. Python has a large ecosystem of libraries for areas such as scientific computing, medical imaging, and machine learning.

A few commonly used Python libraries, some of which we've used already, are:

* **[NumPy](https://numpy.org/doc/stable/user/whatisnumpy.html)** – Numerical computing with arrays and matrices.
* **[SciPy](https://docs.scipy.org/doc/scipy/tutorial/index.html)** – Scientific and mathematical algorithms, such as optimization, statistics, and image processing.
* **[Matplotlib](https://matplotlib.org/)** - Interactive visualizations.
* **[pandas](https://pandas.pydata.org/docs/getting_started/overview.html)** – Working with structured data, especially tabular data.
* **[PyTorch](https://docs.pytorch.org/docs/)** – Optimization and machine learning, particularly with neural networks. Extensively used in the course [AI for Medical Imaging](https://mix.isi.uu.nl/courses/ai-for-medical-imaging/).

These libraries are all 100% open source, and developed by a large community of volunteers. The combination NumPy, SciPy and Matplotlib covers a significant part of the capabilities of MATLAB, and are used in many research projects.

## Dependencies and Package Management
Some Python libraries are included with Python itself, such as `datetime`, while others must be installed separately. Library packages are installed using a *package manager* such as `pip`. `pip` itself is a Python script that downloads packages from a *package repository* (by default <https://pypi.org/>), and installs them in a special packages directory on your computer. The <https://pypi.org/> repository website is also a useful starting point for searching information about Python packages, because it provides links to package documentation, source code, and other resources: see e.g. <https://pypi.org/project/numpy/>.

When you try to import a module from a package that is not installed yet, Python will raise an error such as `ModuleNotFoundError: No module named 'scipy'`. Then simply run:

```shell
pip install scipy
```

or

```shell
python -m pip install scipy
```

from the terminal to install it.

Libraries can be dependent on other libraries: Matplotlib, SciPy, and pandas, for example, all use NumPy for their arrays. As a matter of fact, there are far more libraries that are dependent on other libraries than libraries without any *dependencies* at all. A package manager such as `pip` traces all these dependencies and installs all packages that are required to use the one that you asked for.

You can imagine that things can get complicated when libraries evolve and new functions are added and obsolete ones are removed. A specific *version* of Matplotlib is *compatible* with only certain versions of NumPy. `pip` checks all this and reports an error if there are any incompatible dependency requirements, because it is **not** possible to install multiple versions of the same package in one Python environment.

## Virtual Environments
In the first week of this course, you were asked to install Python and create a so-called *virtual environment*. You can think of a virtual environment as an isolated container for each Python project. Each virtual environment has its own Python interpreter, its own set of installed packages, and is isolated from other virtual environments. This allows us to work on different projects that may have incompatible dependency requirements. It prevents conflicts between projects, makes projects more reproducible, and keeps your system Python installation clean.

A new virtual environment can be created by running e.g.

```shell
python -m venv my_venv
```

from a terminal, where `my_venv` is the name of the virtual environment. Once created, the environment can be activated by running the shell script

```shell
.\my_venv\Scripts\Activate.ps1
```

on Windows, or

```shell
source ./my_venv/bin/activate
```

on macOS or Linux.

When you're working on multiple projects that require some Python coding, you could organize it like this:
```
projects/
├── pmi/
│   ├── .venv/       # Contains e.g. Python 3.14 with numpy 2.5.3, pydicom 3.0.2, etc.
│   ├── images/
│   ├── 2-scripting.ipynb
│   └── ...
├── major/
│   ├── major_venv/  # Contains e.g. Python 3.14 with numpy 2.4.6, torch 2.13.0, etc.
│   ├── tests.ipynb
│   ├── functions.py
│   └── ...
├── research/
│   ├── venv/        # Contains e.g. Python 3.12 with numpy 2.5.3, simpleitk 2.5.6, etc.
│   ├── statistics.py
│   └── ...
└── ...
```
