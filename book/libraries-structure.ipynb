The previous chapter explained that a library is a collection of code that other programs can use. In this chapter we will zoom in on how this 'collection of code' is structured, because once we know that, we can create our own libraries!

## Packages, Modules, and Functions

A package in Python is essentially a directory containing a collection of *modules* (`.py` files), and an `__init__.py` file. When Python encounters a directory with a `__init__.py` file, it treats that directory as a package. The name of the directory is the package name. The code in the `__init__.py` file is executed when the package is imported. The file can be empty, but it could be used to e.g. setup logging or configuration.

As an example, consider the directory:

```
my_package/
├── __init__.py
├── math.py
└── io.py
```

When you execute `import my_package`, then Python will execute the `__init__.py` script. When you need I/O functions from this package, you can execute `import my_package.io`. If you need a specific function, assume e.g. that there is a function `read_dicom()` defined in `io.py`, then you could execute `from my_package.io import read_dicom`.

When you're organizing your code into multiple `.py` files, you'll find that it is sometimes necessary to import a function from another `.py` file. In the example above, a function in `io.py` might, for example, use the function `ras_to_lps()` from `math.py`. In this case you'll need to add `from .math import ras_to_lps` to `io.py`. The `.` tells Python to look for a `math.py` relative to the current directory/package. Note that writing `from math import ras_to_lps` makes Python try to import from the built-in package `math`!

Besides modules, a package can also contain sub-packages, i.e. subdirectories with their own `__init__.py` script and modules. In this way, packages help organize code into manageable, logically separated components; a well-organized package simplifies navigation and understanding of the codebase:

```
my_package/
├── __init__.py
├── conversion.py
├── io.py
└── filters/
    ├── __init__.py
    ├── convolution.py
    ├── morphology.py
    └── ...
```

## An Example: SciPy

Let's have a look at an example of such a well-organized package. The package `SciPy` package provides a collection of mathematical algorithms and convenience functions built on NumPy. These algorithms and functions are organized into [subpackages covering different scientific computing domains](https://docs.scipy.org/doc/scipy/tutorial/index.html). A few very interesting subpackages for image analysis are:

| Subpackage          | Description and User Guide                                                                                     | Examples                                           |
|---------------------|----------------------------------------------------------------------------------------------------------------|----------------------------------------------------|
| `scipy.interpolate` | [Interpolation (scipy.interpolate)](https://docs.scipy.org/doc/scipy/reference/interpolate.html)               | Linear interpolation, B-spline interpolation, etc. |
| `scipy.io`          | [File IO (scipy.io)](https://docs.scipy.org/doc/scipy/reference/io.html)                                       | Reading and writing MATLAB files, etc.             |
| `scipy.linalg`      | [Linear Algebra (scipy.linalg)](https://docs.scipy.org/doc/scipy/reference/linalg.html)                        | Matrix inversion, eigenvalue problems, etc.        |
| `scipy.ndimage`     | [Multidimensional Image Processing (scipy.ndimage)](https://docs.scipy.org/doc/scipy/reference/ndimage.html)   | Convolution, Gaussian filtering, etc.              |
| `scipy.optimize`    | [Optimization (scipy.optimize)](https://docs.scipy.org/doc/scipy/reference/optimize.html)                       | Least-squares fitting, etc.                        |

The `scipy.ndimage` subpackage provides a few modules with various functions for multidimensional image processing. These modules are `filters`, `fourier`, `interpolation`, `measurements`, and `morphology`. The `filters` module contains functions such as `gaussian_filter1d` (for 1-dimensional Gaussian filtering), `gaussian_filter` (for multidimensional Gaussian filtering), and `gaussian_laplace` (multidimensional Laplace filtering using Gaussian derivatives).

You can probably imagine that this is structure is helpful to both users and developers. Library users can easily browse the library to find the functionality they need. Library developers can easily reuse code, because there will be similarities between the functions and algorithms in a module. The `gaussian_filter`, for example, just calls `gaussian_filter1d` for each image dimension, and `gaussian_laplace` uses `gaussian_filter` for its derivatives.

## Namespaces and Scope

By now we've come across a few hierarchical levels of organizing code: a Python installation contains packages, packages contain modules, and modules contain functions. All these levels (packages, modules, functions) have their own *namespace*. The existence of distinct namespaces allows you to have several different instances of a variable or function with the same name. As long as each instance is in a different namespace, they're all maintained separately and won't interfere with one another.

Take, for example, a function named `solve`. NumPy provides a [numpy.linalg.solve](https://numpy.org/doc/stable/reference/generated/numpy.linalg.solve.html), and SciPy also provides a [scipy.linalg.solve](https://docs.scipy.org/doc/scipy/reference/generated/scipy.linalg.solve.html). They can coexist, because both are defined in their own namespaces. The import `from numpy.linalg import solve` binds the NumPy implementation to the current namespace, making the name `solve` available in the current scope. To import both the NumPy *and* SciPy implementations in the current namespace, we need a unique name, which can be achieved in a few different ways:

* `import numpy` - Imports the entire NumPy package. `solve` is available as `numpy.linalg.solve`
* `import numpy as np` - Imports the entire NumPy package and bind it to `np`. `solve` is available as `np.linalg.solve`
* `import numpy.linalg` - Imports the `numpy.linalg` sub-package. `solve` is available as `numpy.linalg.solve`
* `from numpy.linalg import solve as np_solve` - Imports only `solve` and bind it to `np_solve`
