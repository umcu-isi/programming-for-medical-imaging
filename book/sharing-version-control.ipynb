Writing software is a gradual process, similar to writing a thesis, doing research or doing home-improvement. You start with something simple (sometimes from scratch, sometimes by building on someone else's work) and then keep adding small improvements and new features. Some of these new features will be experimental - they might work, or they might fail. You might want to 'undo' your changes and go back to an earlier version of your code.

When doing research, it’s especially important to archive previous versions of your code and documentation. That way, if you need to reproduce your results months or even years later, you’ll know exactly which code and procedures you used.

You might also collaborate with a colleague on the same project. In that case, both of you work on your own copy of the code, and eventually you’ll need to merge your versions back into a single version.

In all these cases, a proper *version control system* is a must-have.

## Git

Nowadays, *Git* is the most frequently used software version control system. Git is basically a smart 'history tracker' for your code and text files. This is what it does:

* **Saves snapshots of your work**: Every time you make changes, you can take a 'snapshot' (called a *commit*). You can go back to that version later if needed.
* **Keeps track of changes**: You can see *who changed what and when*, which is useful when working in a team.
* **Lets you experiment safely**: You can create *branches* (like separate workspaces) to try out new ideas without interfering with the main project. If it works, you merge it back in.
* **Helps teams work together**: Multiple people can work on the same project at the same time, on their own branches, and eventually merge their work.

Note that **Git != GitHub**. *GitHub* is a website for storing and sharing Git repositories. The existance of GitHub is, however, one of the reasons that Git has become the most popular version control system.

Instead of bookkeeping versions of code or documents by creating copies such as `script_v1.py`, `script_v13_final_old.py`, `paper-30-nov-2026.md`, or `data-with-corrections-USE-THIS.xlsx`, Git allows you to archive and access different versions of a files in a much cleaner way.

> * This video nicely explains how Git could be used for bookkeeping academic work: <https://www.youtube.com/watch?v=PBRmtABqC9U>.

*Note that the author mentions that Git only stores differences of changed files. This is wrong: Git **does** store a copy of each changed file, but it applies a compression to keep the repository small.* He does make a valid point, however; when you store large files that frequently change, the repository could become rather large and slow.

Note that this video doesn't even mention GitHub! If you're working on a project on your own, a local repository (i.e. the `.git` directory created by running `git init`) is sufficient for all the bookkeeping. Of course, make sure that this directory is backed up. Only when you would like to share your work with others, or collaborate with others on the same project, then you need some place to store a shared repository: this is what GitHub provides. Many companies and universities also provide local intranet repository sites for bookkeeping code that is not meant to be shared publically.

Copies of Git repositories are called *clones*. This is an appropriate term, because after *cloning* a repository, you have copied everything; all the commits, the branches, the entire history. Actually, in Git there is no such thing as a 'master repository'. All clones (i.e. yours, your colleague's, and the one on GitHub) are exactly the same.

> * This video shows how the Git commands `clone`, `add`, `commit`, `push`, `pull`, and `checkout` work: <https://www.youtube.com/watch?v=e9lnsKot_SQ>

What these videos don't explain is which files should be archived in Git, and which not. The rule is simple: *don't archive anything that was generated from other source files*. So *do* archive your source code, configuration files, scripts, documentation, and assets such as small test-images in Git. *Don't* store virtual environments (just re-create them when needed!), compiled libraries and executables (just re-compile them!), log files, caches, temporary files, etc. Also, *don't* store trained machine learning models or large datasets in Git; these should be stored in other dedicated repositories.
