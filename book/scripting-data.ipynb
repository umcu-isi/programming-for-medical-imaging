## Composite data

A variable can not only hold numbers and text strings, but also more complex, composite data such as dates, tables or images. Working with this kind of data requires *importing* specialized *modules*, such as the `datetime` module:

```python
import datetime

current_time = datetime.datetime.now()
print("This notebook was created on:", current_time)
```

> **External course material**
>
> Review this tutorial about the `datetime` module: <https://www.w3schools.com/python/python_datetime.asp>, and do the exercise: <https://www.w3schools.com/python/python_challenges_datetime.asp>.


The dot (`.`) is used to access functions and member variables of specific modules and composite datatypes. For example, `current_time.year` can be used to read the year from the variable `current_time`:

```python
print("The year was:", current_time.year)
print("Next year is:", current_time.year + 1)
```

Some composite datatypes even allow computation. The `datetime` module, for example, allows calculations with time differences. Note how the module takes leap years into account:

```python
in_ten_years = current_time + datetime.timedelta(days=10 * 365)  # Use the function datetime.timedelta to add 10× 365 days
print("AI has taken over the world in:", in_ten_years)
```

The module also makes sure that all dates and times are valid. Note how it *raises an error* when we try to create an invalid date. The error message traces back the origin of the error, and shows the reason in the last line:

```python
leap_day = datetime.datetime(2024, 2, 29)
not_a_leap_day = datetime.datetime(2026, 2, 29)
```

## Reading and writing data
Variables allow us to do computations and manipulate complex data. But once a computer program quits, the variables and their values, which are stored in the computer's *volatile* memory, are gone. In order to preserve the information, we need to write and read the data to *non-volatile storage* such as a local-, network-, or cloud-storage drive.

A very frequently used Python module to work with tabular data (spreadsheets, tables) is [pandas](https://pandas.pydata.org/docs/getting_started/overview.html). The `pandas` module allows creating, manipulating, reading, and writing various standard tabular *file formats*, such as Excel, SPSS, and CSV. Note that saving a file in such a standard format allows transferring data between Python and other programs.

The code below shows a simple example of how `pandas` can be used to create, write, and read a table.

```python
import pandas as pd

# Create a simple pandas DataFrame (a table).
tissue_volumes = pd.DataFrame({"Volume": [723, 512, 183]},
                              index=["Gray matter", "White matter", "CSF"])
tissue_volumes.index.name = "Tissue type"

# Save the DataFrame to a CSV file.
tissue_volumes.to_csv("tissue_volumes.csv")
```

```python
# Load the CSV file and display the table.
loaded_volumes = pd.read_csv("tissue_volumes.csv", index_col="Tissue type")
loaded_volumes
```

```python
# Lookup the gray matter volume and convert it to liters.
tissue = "Gray matter"
print(tissue, ":", loaded_volumes.loc[tissue, "Volume"] / 1000, "L")
```
