## Composite data

A variable can not only hold numbers and text strings, but also more complex, composite data such as dates, tables or images. Working with this kind of data requires *importing* specialized *modules*, such as the `datetime` module:

```python
import datetime

timestamp = datetime.datetime.now()
print("This notebook was created on:", timestamp)
```

The dot (`.`) is used to call functions and access variables of specific modules and composite datatypes. For example, `timestamp.year` can be used to read the year from the variable `timestamp`:

```python
print("The year was:", timestamp.year)
print("Next year is:", timestamp.year + 1)
```

Some composite datatypes even allow computation. The `datetime` module, for example, allows calculations with time differences:

```python
in_ten_years = timestamp + datetime.timedelta(days=10 * 365)  # Use the function datetime.timedelta to add 10× 365 days to 'timestamp'
print("AI has taken over the world in:", in_ten_years)
```
