The pixel values of a CT image are generally stored as 12-bit integer numbers, ranging from 0 to ${2^{12}}-1=4095$. They represent, however, CT values (in [Hounsfield units](https://en.wikipedia.org/wiki/Hounsfield_scale)), ranging from -1000 HU (air) to 0 HU (water) and higher.
